Run all cells, then open the printed `trycloudflare.com` URL (Cloudflare quick

```python
# 1) Clone vid2rig from GitHub (prebuilt web/dist included, no npm needed)
import os

REPO_URL = "https://github.com/liladhiee-web/vid2rig.git"
REPO_DIR = "/kaggle/working/vid2rig"

if not os.path.isdir(os.path.join(REPO_DIR, "server")):
    if os.path.exists(REPO_DIR):  # partial clone — start clean
        import shutil
        shutil.rmtree(REPO_DIR)
    os.environ["GIT_TERMINAL_PROMPT"] = "0"
    !git clone --depth 1 {REPO_URL} {REPO_DIR}
else:
    print("Repo already present at", REPO_DIR)

os.environ["VID2RIG_DIR"] = REPO_DIR
assert os.path.isdir(os.path.join(REPO_DIR, "server")), "Clone failed — check Internet setting"
print("Repo ready:", REPO_DIR)
```

```python
# 2) Install deps + start backend + open Cloudflare tunnel
import sys
sys.path.insert(0, "/kaggle/working/vid2rig/kaggle")
from kaggle_bootstrap import bootstrap
bootstrap()
```

```python
# 3) Keep-alive: run this cell while using the app (shows health every 60s)
import time, requests
while True:
    try:
        h = requests.get("http://localhost:8000/api/health", timeout=5).json()
        print(h, flush=True)
    except Exception as e:
        print("backend down:", e, flush=True)
    time.sleep(60)
```
