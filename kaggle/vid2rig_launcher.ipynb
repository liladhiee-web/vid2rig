# vid2rig launcher (Kaggle free GPU)

**Setup before running:**
1. Settings → Accelerator → **GPU T4 x2**
2. Settings → Internet → **On**
3. Attach a **private dataset** containing `SMPL_NEUTRAL.pkl` (register free at smpl-x.is.tue.mpg.de)
4. The repo URL (`github.com/liladhiee-web/vid2rig`) is already set in the first cell — no dataset needed

Run all cells, then open the printed `trycloudflare.com` URL (Cloudflare quick
tunnel — free, no account or token needed).

```python
# Clone vid2rig from GitHub (prebuilt web/dist included, no npm needed)
import os
os.environ["VID2RIG_REPO_URL"] = "https://github.com/liladhiee-web/vid2rig.git"
```

```python
# Install + start backend + open tunnel
import sys
sys.path.insert(0, "/kaggle/usr/lib")  # if uploaded as a Kaggle utility script
from kaggle_bootstrap import bootstrap
bootstrap()
```

```python
# Keep-alive: run this cell while using the app (shows tunnel URL again)
import requests
while True:
    try:
        h = requests.get(f"http://localhost:8000/api/health", timeout=5).json()
        print(h, flush=True)
    except Exception as e:
        print("backend down:", e, flush=True)
    import time; time.sleep(60)
```
